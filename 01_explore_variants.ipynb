```python
df = pd.read_csv(
    "data/clinvar_download_20261002_085509.tsv",
    sep="\t",
    index_col=False
)

print("Total variants:", len(df))
print()
for i, col in enumerate(df.columns):
    print(i, "|", col, "->", df[col].iloc[0])
```

```python
print("Total variants:", len(df))
print()
print("=== Germline classification ===")
print(df["Germline classification"].value_counts(dropna=False))
print()
print("=== Variant type ===")
print(df["Variant type"].value_counts(dropna=False))
print()
print("=== Molecular consequence (raw) ===")
print(df["Molecular consequence"].value_counts(dropna=False))
```

```python
def classify_consequence(mc):
    """Assign each variant to a single primary consequence class,
    using a severity hierarchy (ClinVar lists multiple consequences
    per variant across transcripts)."""
    if pd.isna(mc):
        return "not_specified"
    mc = mc.lower()
    if "frameshift" in mc:
        return "frameshift"
    if "nonsense" in mc:
        return "nonsense"
    if "splice" in mc:
        return "splice"
    if "inframe" in mc:
        return "inframe_indel"
    if "initiator_codon" in mc or "start_lost" in mc:
        return "start_loss"
    if "missense" in mc:
        return "missense"
    if "synonymous" in mc:
        return "synonymous"
    if "utr" in mc:
        return "utr"
    if "intron" in mc:
        return "intron"
    return "other"

df["consequence_class"] = df["Molecular consequence"].apply(classify_consequence)

print(df["consequence_class"].value_counts())
print()
print("Total:", df["consequence_class"].value_counts().sum())
```

```python
print(pd.crosstab(df["consequence_class"], df["Variant type"]))
```

```python
# Classify each variant by whether a single-base editing approach
# could, in principle, address it.
# This is a PERMISSIVE upper bound: it ignores PAM availability,
# editing window position, and bystander edits.

def editability_upper_bound(row):
    """Can a base editor (CBE/ABE) in principle correct this variant?
    Only single-nucleotide changes are candidates; any insertion or
    deletion changes sequence length and is structurally out of reach."""
    if row["Variant type"] != "single nucleotide variant":
        return "out_of_reach_indel"
    if row["consequence_class"] in ("missense", "nonsense", "splice", "intron"):
        return "candidate_SNV"
    return "other"

df["be_upper_bound"] = df.apply(editability_upper_bound, axis=1)

counts = df["be_upper_bound"].value_counts()
print(counts)
print()
for k, v in counts.items():
    print(f"{k}: {v} ({100*v/len(df):.1f}%)")
```

```python
# Within the SNV candidates, which base-editing chemistry would be needed?
snv = df[df["be_upper_bound"] == "candidate_SNV"].copy()

print("SNV candidates by consequence:")
print(snv["consequence_class"].value_counts())
print()

# Extract the nucleotide change from the Name field, e.g. "c.1601A>T"
import re

def extract_sub(name):
    m = re.search(r"c\.[\d\-\+\*_]+([ACGT])>([ACGT])", str(name))
    if m:
        return m.group(1) + ">" + m.group(2)
    return None

snv["substitution"] = snv["Name"].apply(extract_sub)

print("Substitution types found:", snv["substitution"].notna().sum(), "of", len(snv))
print()
print(snv["substitution"].value_counts(dropna=False))
```

```python
# A base editor can only install transitions:
#   ABE: A->G (and T->C on the opposite strand)
#   CBE: C->T (and G->A on the opposite strand)
# To CORRECT a variant, the required change is the reverse of the
# substitution that caused it.

correctable_by_BE = {
    "G>A": "ABE",   # needs A->G
    "C>T": "ABE",   # needs T->C
    "A>G": "CBE",   # needs G->A
    "T>C": "CBE",   # needs C->T
}

snv["BE_chemistry"] = snv["substitution"].map(correctable_by_BE).fillna("none")

print(snv["BE_chemistry"].value_counts())
print()

n_correctable = (snv["BE_chemistry"] != "none").sum()
print(f"SNVs chemically correctable by a base editor: {n_correctable} / {len(snv)}")
print(f"As a fraction of ALL pathogenic variants: {n_correctable} / {len(df)} = {100*n_correctable/len(df):.1f}%")
```

```python
def final_class(row):
    if row["Variant type"] != "single nucleotide variant":
        return "Indel"
    sub = extract_sub(row["Name"])
    if sub in ("G>A", "C>T"):
        return "ABE"
    if sub in ("A>G", "T>C"):
        return "CBE"
    return "Transversion"

df["coverage_class"] = df.apply(final_class, axis=1)
print(df["coverage_class"].value_counts())
```

```python
import matplotlib.pyplot as plt

order = ["ABE", "CBE", "Transversion", "Indel"]
vals = df["coverage_class"].value_counts().reindex(order).values
pct = 100 * vals / len(df)
colors = ["#2E7D32", "#66BB6A", "#EF6C00", "#B71C1C"]

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(order, vals, color=colors, edgecolor="black")
ax.set_ylabel("Pathogenic variants")
ax.set_title("IDS variants accessible to base editing")
ax.set_ylim(0, 200)
plt.tight_layout()
plt.savefig("results/fig1_coverage.png", dpi=300)
plt.show()
```

```python
import re

def aa_position(pc):
    m = re.search(r"[A-Z](\d+)", str(pc))
    return int(m.group(1)) if m else None

df["aa_pos"] = df["Protein change"].apply(aa_position)

print("Most frequently hit codons:")
print(df["aa_pos"].value_counts().head(15))
```

```python
total_codons = df["aa_pos"].notna().sum()
unique_codons = df["aa_pos"].nunique()

print(f"Variants with codon info: {total_codons}")
print(f"Distinct codons affected: {unique_codons}")
print(f"Mean variants per codon: {total_codons/unique_codons:.2f}")
print()

top = df["aa_pos"].value_counts()
print(f"Top codon accounts for: {top.iloc[0]}/{total_codons} = {100*top.iloc[0]/total_codons:.1f}%")
print(f"Top 5 codons account for: {top.head(5).sum()}/{total_codons} = {100*top.head(5).sum()/total_codons:.1f}%")
print(f"Codons hit by only ONE variant: {(top==1).sum()} ({100*(top==1).sum()/unique_codons:.1f}% of codons)")
```

```python
def aa_position_fixed(pc):
    """Extract codon position from the canonical transcript.
    ClinVar lists positions for two isoforms in inconsistent order;
    the full-length NM_000202.8 numbering is the larger value."""
    if pd.isna(pc):
        return None
    nums = re.findall(r"[A-Z](\d+)", str(pc))
    if not nums:
        return None
    return max(int(n) for n in nums)

df["aa_pos"] = df["Protein change"].apply(aa_position_fixed)

print("Range:", df["aa_pos"].min(), "-", df["aa_pos"].max())
print("Distinct codons:", df["aa_pos"].nunique())
print()
print(df["aa_pos"].value_counts().head(10))
```

```python
def cdna_pos(name):
    m = re.search(r"c\.(\d+)", str(name))
    return int(m.group(1)) if m else None

df["cdna"] = df["Name"].apply(cdna_pos)
print("cDNA range:", df["cdna"].min(), "-", df["cdna"].max())
print()
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df["cdna"].dropna(), bins=50, color="#1565C0", edgecolor="black")
ax.set_xlabel("cDNA position (c.)")
ax.set_ylabel("Variants")
plt.tight_layout()
plt.show()
```
